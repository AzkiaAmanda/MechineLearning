setelah dicek ternyata ada missing value pada kolom K. maka dilakukan penghapusan data kosong pada dataset tersebut menggunakan fungsi df.dropna dengan pengaturan subset 'K'. dilanjut dengan pengecekan ulang missing value pada dataset.

```python
df_1 = df.dropna(subset=['K'])
missing_value_after_drop = df_1.isnull().sum()
print(missing_value_after_drop)
```

Membuat tabel korelasi dengan fungsi .corr dan menentukan kolom apa saja yang akan dicari korelasinya (kolomnya itu bisa dimasukkan dari keseluruhan kolom target dan fitur)

```python
corr_matrix = df[['K', 'b5', 'b7', 'b11', 'Sigma_VV', 'plia']].corr()
print(corr_matrix)
```

Menentukan fitur yang dibuat dalam variabel X dan target dibuat dalam variabel y.
pembagian dataset menjadi data testing sebanyak 20% atau 119 data dan data training sebanyak 80% atau 474 data.

```python
from sklearn.model_selection import train_test_split

X = df_1[['b5', 'b7', 'b11', 'Sigma_VV', 'plia']]
y = df_1[['K']]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=7
)

#cetak data training
print("Jumlah data train:", len(X_train))
print("Jumlah data test :", len(X_test))
X_train.head()
```

memvisualisasikan korelasi menjadi heatmap dengan pengaturan ukuran 9,6

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Heatmap Korelasi Dataset Satelit terhadap K")
plt.show()
```

dilakukan penambahan kolom konstanta untuk selanjutnya pembuatan model ols.

```python
import statsmodels.api as sm

X_train_const = sm.add_constant(X_train)
X_train_const.head()
```

tahap ini membuat model regresi menggunakan sm.OLS, dan melihat nilai koefisien menggunakan model.params.
agar lebih tergambar maka dilakukan pencetakan persamaan regresi.

```python
# Buat model OLS
model = sm.OLS(y_train, X_train_const).fit()
print('-------------------------')
print(model.params)
print('-------------------------')
const = model.params['const']
x1 = model.params['b5']
x2 = model.params['b7']
x3 = model.params['b11']
x4 = model.params['Sigma_VV']
x5 = model.params['plia']
#print persamaan regresi
print(f"y = {const:.3f} + {x1:.3f}*b5 + {x2:.3f}*b7 + {x3:.3f}*b11 + {x4:.3f}*Sigma_VV + {x5:.3f}*plia")
```

mencetak hasil model regresi dengan pemanggilan model dan dilanjut dengan pemanggilan fungsi .summary.

```python
print(model.summary())
```

dilakukan penambahan nilai konstanta pada data testing. lalu dilakukan prediksi target (y) menggunakan model hasil training.
membuat tabel hasil prediksi yang disimpan pada variabel hasil. pada tahap ini .to_numpy ini berfungsi untuk mengubah data menjadi bentuk array numpy.
menghitung error dengan mengurangi hasil dari prediksi target dengan data aktual target.
terakhir dilakukan perhitungan persentasi akurasi.

```python
import numpy as np

#Tambahkan konstanta ke data uji
X_test_const = sm.add_constant(X_test)
#Prediksi
y_pred_test = model.predict(X_test_const)
#Buat tabel hasil prediksi
hasil = pd.DataFrame({
    "Red Edge": X_test["b5"].to_numpy(),
    "SWIR(Short Wave Infrared)": X_test ["b7"].to_numpy(),
    "SWIR 2 (Short Wave Infrared)": X_test ["b11"].to_numpy(),
    "Radar Backscatter": X_test ["Sigma_VV"].to_numpy(),
    "Sudut Lokal": X_test ["plia"].to_numpy(),
    "Kadar Kalium Aktual": y_test.to_numpy().flatten(),
    "Kadar Kalium Prediksi": y_pred_test.to_numpy().flatten()
})
#1) Selisih error (positif overpredict)
hasil["Selisih error"] = hasil ["Kadar Kalium Prediksi"] - hasil ["Kadar Kalium Aktual"]

#2) Akurasi per-baris (100 (1/error//aktual)), dibatasi 0-100
denom = hasil ["Kadar Kalium Aktual"].replace(0, np.nan) # antisipasi pembagi nol
hasil ["Akurasi (%)"] = (1 - (hasil ["Selisih error"].abs() / denom)).clip(lower=0, upper=1) * 100
hasil.head()
```

mengimport library yang digunakan untuk perhitungan tambahan.
MAE: Rata-rata selisih absolut	Kecil
MSE:	Rata-rata kuadrat error	Kecil
RMSE:	Akar kuadrat MSE (lebih mudah dibaca)	Kecil
R²:	Variasi data yang dijelaskan model

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

mse = mean_squared_error(y_test, y_pred_test)
print(f"Mean Squared Error (MSE): {mse:.4f}")

mae = mean_absolute_error(y_test, y_pred_test)
print(f"Mean Absolute Error (MSE): {mae:.4f}")

r2 = r2_score(y_test, y_pred_test)
print(f"R-squared (R2): {r2:.4f}")

rmse = np.sqrt(mse)
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
```
